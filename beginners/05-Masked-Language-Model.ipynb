```python
from transformers import BertTokenizer, BertForMaskedLM
import torch
```

```python
# Carga un tokenizador preentrenado compatible con BERT.
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")

# Carga un modelo de BERT utilizando Masked Language Model
model = BertForMaskedLM.from_pretrained("bert-base-uncased")
```

**BertTokenizer**: clase especializada para tokenizar texto como lo espera BERT  
**.from_pretrained()**: descarga automaticamente el modelo del tokenizador desde Hugging Face  
**"bert-base-uncased"**:   
- base: 12 capas, 768 dimensiones ocultas
- uncased: todo el texto se convierte en minusculas
 
**Resultado**:
- Se puede dividir el texto en WordPieces (subpalabras) 
- asignar IDs numericas a cada token


**BertForMaskedLM**: clase que agrega a BERT una capa de salida Softmax para predecir tokens enmascarados
**.from_pretrained()**: descarga pasos entrenados del modelo desde Hugging Face

**Resultado**:
- El modelo puede procesar secuencias tokenizadas con mascaras
- Predecir cual era el token oculto en cada [MASK]

```python
# Frase con máscara
text = "The capital of France is [MASK]."
inputs = tokenizer(text, return_tensors="pt")
mask_token_index = torch.where(inputs["input_ids"] == tokenizer.mask_token_id)[1]
```

```python
# Predicción
with torch.no_grad():
    outputs = model(**inputs)
    logits = outputs.logits
```

```python
mask_word_logits = logits[0, mask_token_index, :]
top_5 = torch.topk(mask_word_logits, 5, dim=1).indices[0].tolist()
```

```python
print("Posibles predicciones:")
for token in top_5:
    print(f" - {tokenizer.decode([token])}")
```
