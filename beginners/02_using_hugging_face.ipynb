```python
# To use the pipeline from transformers
from transformers import pipeline
```

```python
# To create the pipeline
qa = pipeline(
    "question-answering",
    model="distilbert/distilbert-base-cased-distilled-squad",
    framework="pt"  # <- Forzar uso de PyTorch
)

# To use the pipeline
result = qa(
    question="What is Hugging Face?",
    context="Hugging Face is a company."
)

print("Respuesta:", result["answer"])
```

```python
# Crea un pipeline de respuesta a preguntas usando un modelo preentrenado
demo = pipeline("question-answering")

# Contexto para realizar preguntas
context = (
    "Hugging Face es una empresa que ha creado una de las plataformas más importantes para compartir "
    "y utilizar modelos de inteligencia artificial open source. Facilita el acceso a modelos como BERT, GPT, T5, y más."
)

# Preguntas de ejemplo
questions = [
    "¿Qué es Hugging Face?",
    "¿Qué modelos ofrece Hugging Face?",
    "¿Para qué sirve Hugging Face?"
]

# Ejecuta las preguntas sobre el contexto y muestra las respuestas
for q in questions:
    result = demo(question=q, context=context)
    print(f"Pregunta: {q}")
    print(f"Respuesta: {result['answer']}\n")
```

```python
# To use AutoTokenizer and AutoModel

from transformers import AutoTokenizer, AutoModel
```

```python
# Cargar el tokenizer y el modelo preentrenado (BERT sin clasificación)
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased")

# Texto de entrada
text = "Hugging Face is changing the way we build AI."

# Tokenizar el texto
inputs = tokenizer(text, return_tensors="pt")  # pt = PyTorch

# Mostrar tokens y tensores
print("Tokens:", tokenizer.tokenize(text))
print("Input IDs:", inputs["input_ids"])
print("Attention mask:", inputs["attention_mask"])

# Pasar los tokens por el modelo
with torch.no_grad():  # Desactiva el cálculo de gradientes (más rápido)
    outputs = model(**inputs)

# Salida: tupla con embeddings
last_hidden_state = outputs.last_hidden_state  # shape: (batch_size, sequence_length, hidden_size)
print("Last hidden state shape:", last_hidden_state.shape)
```
